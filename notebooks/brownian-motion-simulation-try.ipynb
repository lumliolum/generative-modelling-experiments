```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
# 0 to 10
T = 1
# samples1
nsamples = 1024

delT = T / nsamples

samples1 = [0]
for i in range(1, nsamples):
    # generate a random with variance as delta-T
    # so we generate the increment and then add it
    samples1.append(samples1[-1] + np.random.normal(0, np.sqrt(delT)))

samples1 = np.array(samples1)
print(samples1.shape)

plt.figure(figsize=(15, 5))
plt.plot(samples1)
```

```python
# 0 to 10
T = 1
# samples2
nsamples = 1024

delT = T / nsamples

samples2 = [0]
for i in range(1, nsamples):
    # generate a random with variance as delta-T
    # so we generate the increment and then add it
    samples2.append(samples2[-1] + np.random.normal(0, np.sqrt(delT)))

samples2 = np.array(samples2)
print(samples2.shape)

plt.figure(figsize=(15, 5))
plt.plot(samples2)
```
