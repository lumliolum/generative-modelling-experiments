```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
train_df = pd.read_csv("data/mnist_train.csv", header=None)
print(train_df.shape)
train_df.head()
```

```python
# conver to numpy array
X = train_df.values
X = X / 255.0
print(X.shape)
```

```python
def perform(X):
    index = 475
    nsamples = 1000
    # get the selected row
    row = X[index: index + 1, :]
    distances = np.sum((row - X)**2, axis=1)

    closest_indices = np.argsort(distances)[1 : nsamples + 1]
    # print the radius
    print(distances[1 : nsamples + 1][-1])

    mat = (X[closest_indices] - row)
    # print the rank of matrix
    print(np.linalg.matrix_rank(mat))
```

```python
perform(X)
```
